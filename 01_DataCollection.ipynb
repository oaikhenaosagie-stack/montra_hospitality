```python
## Handling missing values in guest table

guests['loyalty_status'] = guests['loyalty_status'].fillna('Unknown')
```

```python
guests.isnull().mean()*100
```

```python
## Merging all tables.


# Build master review dataset
rev_df = (
    reviews
    .merge(
        bookings,
        on=['booking_id', 'guest_id', 'hotel_id'],
        how='left'
    )
    .merge(
        guests,
        on='guest_id',
        how='left'
    )
    .merge(
        hotels,
        on='hotel_id',
        how='left'
    )
)

print(rev_df.shape)
rev_df.head()
```

##### Notes

- Tables were extracted from the database and inspected to identify matching columns and presence of missing values before merging them together.

- Identified missing values (greater than 50%) in loyalty_status column of the guests table was replaced by "unknown" as the column may be important in downline analysis.

- A left join on the reviews table was used to merge all the tables into one as the reviews table has matching columns in the other tables and the reviews table is the table of interest in sentiment analysis.
